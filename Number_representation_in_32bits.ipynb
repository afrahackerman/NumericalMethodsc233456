```python
def decimal_to_binary_integer(n):
    binary = ""
    if n == 0:
        return "0"
    while n > 0:
        binary = str(n % 2) + binary
        n = n // 2

    return binary


def fp_32(x):
    if x < 0:
        unsigned = "1"
        x = -x
    else:
        unsigned = "0"


    integer_part = int(x)
    fraction_part = x - integer_part

    integer_binary = decimal_to_binary_integer(integer_part)
    fraction_binary = ""

    for i in range(30):
        fraction_part = fraction_part * 2

        if fraction_part >= 1:
            fraction_binary += "1"
            fraction_part -= 1
        else:
            fraction_binary += "0"


    exponent = len(integer_binary) - 1

    mantissa_binary = integer_binary[1:] + fraction_binary #000111=111
    mantissa = mantissa_binary[:23]

    while len(mantissa) < 23:
        mantissa += "0"

    biased_exponent = exponent + 127 #(bias)

    exponent_bits = decimal_to_binary_integer(biased_exponent)

    while len(exponent_bits) < 8:
        exponent_bits = "0" + exponent_bits

    representation = unsigned + exponent_bits + mantissa
    print("Number:", -97.123)
    print("unsigned:", unsigned)
    print("exponent:", exponent_bits)
    print("mantissa:", mantissa)
    print("32-bit representation:", representation)

fp_32(-97.123)
```
