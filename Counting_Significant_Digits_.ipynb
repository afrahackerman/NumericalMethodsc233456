```python
def count_sd(number):
    number = number.strip()

    # Remove sign
    if number.startswith(("+", "-")):
        number = number[1:]

    # Scientific notation not supported
    if "e" in number.lower():
        print("Scientific notation is not supported.")
        return

    # Decimal number
    if "." in number:
        integer_part, decimal_part = number.split(".")

        integer_part = integer_part.lstrip("0")

        if integer_part:
            # Example: 456.7800
            count = len(integer_part + decimal_part)
        else:
            # Example: 0.001230
            decimal_part = decimal_part.lstrip("0")
            count = len(decimal_part) if decimal_part else 1

    # Whole number
    else:
        number = number.lstrip("0")

        if not number:
            count = 1
        else:
            # Trailing zeros are not significant
            count = len(number.rstrip("0"))

    return count


# MAIN
number = input("Enter a number: ")
result = count_sd(number)
print("Number of significant digits:", result)
```
