```python
def bankers_round(number, n):
    number = str(number).strip()

    # Remove sign
    sign = ""
    if number.startswith(("+", "-")):
        sign = number[0]
        number = number[1:]

    # Split number
    if "." in number:
        integer_part, decimal_part = number.split(".")
    else:
        integer_part = number
        decimal_part = ""

    # Remove leading zeros to find significant digits
    digits = (integer_part + decimal_part).lstrip("0")

    if not digits:
        return "0"

    # Nothing to round
    if len(digits) <= n:
        return sign + number

    # Significant digits and deciding digit
    sig = list(digits[:n])
    next_digit = int(digits[n])
    remaining = digits[n + 1:]
    last_digit = int(sig[-1])

    # Decide rounding
    if next_digit > 5:
        round_up = True
    elif next_digit < 5:
        round_up = False
    else:
        # 5 exactly -> banker's rounding
        round_up = any(d != "0" for d in remaining) or last_digit % 2 == 1

    # Round up
    if round_up:
        i = n - 1

        while i >= 0 and sig[i] == "9":
            sig[i] = "0"
            i -= 1

        if i >= 0:
            sig[i] = str(int(sig[i]) + 1)
        else:
            sig.insert(0, "1")

    rounded = "".join(sig)

    # Put decimal point back
    integer_digits = len(integer_part.lstrip("0"))

    if integer_digits:
        if n < integer_digits:
            result = rounded + "0" * (integer_digits - n)
        elif n == integer_digits:
            result = rounded
        else:
            result = rounded[:integer_digits] + "." + rounded[integer_digits:]
    else:
        # Number is less than 1
        zeros = len(decimal_part) - len(decimal_part.lstrip("0"))
        result = "0." + "0" * zeros + rounded

    return sign + result


# MAIN
number = input("Enter a number: ")
n = int(input("Enter number of significant figures: "))

if n <= 0:
  print("Number of significant figures must be greater than 0.")
else:
  result = bankers_round(number, n)
  print(result)
```
